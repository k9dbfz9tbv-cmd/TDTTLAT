### W4A1 - SumOfNumbers


```python
n = int(input())
s = 0
for i in range(1, n + 1):
    s += i
print(s)
```

### W4A2 - PrimeNumber


```python
n = int(input())
while n <= 0:
    n = int(input())

prime = n > 1
for i in range(2, int(n ** 0.5) + 1):
    if n % i == 0:
        prime = False
        break
print("Yes" if prime else "No")
```

### W4A3 - Factorial


```python
n = int(input())
f = 1
for i in range(1, n + 1):
    f *= i
print(f)
```

### W4A4 - CountingDigits


```python
n = abs(int(input()))
d = 1
while n >= 10:
    n //= 10
    d += 1
print(d)
```

### W4A5 - MeaningOfLife


```python
n = int(input())
a = []
while len(a) < n:
    b = list(map(int, input().split()))
    a.extend(b)
print("I've found the meaning of life!" if 42 in a[:n] else "It's a joke!")
```

### W4A6 - SumPrimesInRange


```python
f = [True] * 1000001
f[0]=f[1]=False
for i in range(2, int(1000000**0.5)+1):
    if (f[i]):
        k=2
        while (k*i<=1000000):
            f[k*i] = False
            k+=1

while True:
    try:
        s = input()
    except EOFError:
        break
    if not s:
        break
    a, b = map(int, s.split())
    total = 0
    for n in range(a, b + 1):
        if f[n]:
            total+=n
    print(total)
```

### W4A7 - LargestPrimeFactor


```python
n = int(input())
d = 2
largest = 1
while d * d <= n:
    while n % d == 0:
        largest = d
        n //= d
    d += 1
if n > 1:
    largest = n
print(largest)
```

### W4A8 - ReverseAddPalindrome


```python
n = int(input())
steps = 0
while str(n) != str(n)[::-1]:
    n += int(str(n)[::-1])
    steps += 1
print(steps, n)
```

### W4A9 - PerfectSquaresDistinctDigits


```python
n = int(input())
a = []
i = 1
while i * i <= n:
    x = str(i * i)
    if len(x) == len(set(x)):
        a.append(x)
    i += 1
print(' '.join(a) if a else 'no number')
```

### W4A10 - LongestCollatzUpToN


```python
n = int(input())
best, longest = 1, 0
for i in range(1, n + 1):
    x, length = i, 1
    while x != 1:
        x = x // 2 if x % 2 == 0 else 3 * x + 1
        length += 1
    if length > longest:
        best, longest = i, length
print(best, longest)
```

### W4A11 - EvenDivisors


```python
n = int(input())
count = 0
for i in range(2, n + 1, 2):
    if n % i == 0:
        count += 1
print(count)
```

### W4A12 - Interest


```python
x, n = map(int, input().split())
for _ in range(n):
    x *= 1.007
print(int(x))
```

### W4A13 - ClosePairNumber


```python
a, b = map(int, input().split())
s1 = s2 = 0
for i in range(1, a):
    if a % i == 0:
        s1 += i
for i in range(1, b):
    if b % i == 0:
        s2 += i
print('true' if a != b and s1 == b and s2 == a else 'false')
```

### W4A14 - GCD

```python
m, n = map(int, input().split())
while n:
    m, n = n, m % n
print(m)
```

### W4A15 - ChickenAndDog


```python
heads, legs = map(int, input().split())
dogs = (legs - 2 * heads) // 2
chickens = heads - dogs
if chickens >= 0 and dogs >= 0 and 2 * chickens + 4 * dogs == legs:
    print(chickens, dogs)
else:
    print('invalid')
```
