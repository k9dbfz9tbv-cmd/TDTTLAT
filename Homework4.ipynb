#### W4A20 Nhập vào một số nguyên dương n, kiểm tra xem n có phải là số dạng 2^k hay không

```python
n = int(input())
if n <= 0:
    print("NO")
else:
    while n % 2 == 0:
        n //= 2
    if n == 1:
        print("YES")
    else:
        print("NO")
```

####  W4A22 Nhập vào số nguyên dương n, đếm xem n có bao nhiêu chữ số chẵn, bao nhiêu chữ số lẻ 

```python
n = int(input())
chan = le = 0
while n != 0:
    c = n % 10
    if c % 2 == 0:
        chan+=1
    else: le+=1
    n//=10
print(f"Có {chan} số chẵn và {le} số lẻ")
```

####  W4A24 Nhập vào A, tìm n nhỏ nhất sao cho 1 + 1/2 + 1/3 + 1/4 + ... + 1/n > A

#### W4A26 Dãy số fibonacci là dãy số được định nghĩa như sau: 1, 1, 2, 3, 5, 8, 13,... với số kế tiếp sẽ bằng tổng hai số trước đó Nhập vào số nguyên dương A, hãy tìm số trong dãy số fibonacci lớn nhất nhưng không vượt quá A 

```python
a = int(input())
x = 1
y = 1
while y <= a:
    x, y = y, x + y
print(x)
```

#### W4A27 Nhập vào một chuỗi, hãy đếm xem trong chuỗi có bao nhiêu từ (quy định là chuỗi không có ký tự đặc biệt, không số, không có dấu câu, chỉ có ký tự chữ và khoảng trắng) Dùng cách: for trên index; KHÔNG dùng hàm split(), len()

```python
n = input()
m = 0
for x in n:
    m += 1
cnt = 0
for i in range(m):
    if n[i] != ' ' and (i == 0 or n[i-1] == ' '):
        cnt += 1
print(cnt)
```

#### W4A28 Nhập vào một chuỗi, hãy in từ đầu tiên trong chuỗi. Dùng cách: for trên index; KHÔNG dùng hàm split() 

```python

s = input()
res = ""
bat_dau = False
for i in range(len(s)):
    if s[i] != ' ':
        res += s[i]
        bat_dau = True
    elif bat_dau:
        break
print(res)
```

#### W4A29 Cho một chuỗi chỉ gồm các ký tự chữ cái thường và chữ cái hoa, hãy đổi ký tự hoa thành thường và thường thành hoa. 

```python
s = input()
res=""
for i in s:
    if i.isupper(): res+=i.lower()
    else: res+=i.upper()
print(res)
```

#### W4A30 Nhập vào một chuỗi có dạng 3 số nguyên, mỗi số nguyên cách nhau một dấu phẩy, hãy tính tổng 3 số nguyên đó 

```python
s = input()
i = 0
res = 0
while i < len(s):
    if s[i] == ',' or s[i] ==  ' ':
        i+=1
    else:
        check = ''
        j = 0
        while i+j<len(s) and s[i+j].isdigit():
            check+=s[i+j]
            j+=1
        res += int(check)
        i+=j
    # i+=len(cnt)
print(res)
```

#### W4A31 Nhập vào một chuỗi, hãy đếm xem trong chuỗi có bao nhiêu ký tự in hoa, bao nhiêu ký tự in thường, bao nhiêu ký tự số 

```python
s = input()
hoa = thuong = so = 0
for i in s:
    if i.isupper(): hoa+=1
    elif i.islower(): thuong+=1
    elif i.isdigit(): so+=1
print(f"Có {hoa} ký tự hoa, {thuong} ký tự thường, {so} kí tự số")
```
