### W3A1 – Đảo ngược số có 3 chữ số (không dùng chuỗi)

```python
n = int(input())
if 100 <= n <= 999:
    print(n % 10 * 100 + n // 10 % 10 * 10 + n // 100)
```

### W3A2 – Hoán đổi hai số bằng XOR

```python
a, b = map(int, input().split())
a ^= b
b ^= a
a ^= b
print(a, b)
```

### W3A3 – Kiểm tra lũy thừa của 2 bằng toán tử bit

```python
n = int(input())
print(n > 0 and (n & (n - 1)) == 0)
```

### W3A4 – Chia lấy kết quả làm tròn xuống

```python
m, n = map(int, input().split())
print(m // n)
```

### W3A5 – Chia lấy kết quả làm tròn lên

```python
m, n = map(int, input().split())
print(-(-m // n))
```

### W3A6 – Even and Odd

```python
x = int(input())
print("Even" if x % 2 == 0 else "Odd")
```

### W3A7 – Check negative numbers

```python
a, b = map(int, input().split())
print("Yes" if a < 0 and b < 0 else "No")
```

### W3A8 – Longer string

```python
a, b = input().split()
print(len(a) > len(b))
```

### W3A9 – Triangle

```python
a, b, c = map(int, input().split())
print("Yes" if a + b > c and a + c > b and b + c > a else "No")
```

### W3A10 – Largest number

```python
a, b, c, d = map(int, input().split())
print(max(a, b, c, d))
```

### W3A11 – Phân loại tam giác

```python
a, b, c = map(int, input().split())
if a + b <= c or a + c <= b or b + c <= a:
    print("Không phải tam giác")
elif a == b == c:
    print("Tam giác đều")
elif a == b or b == c or a == c:
    print("Tam giác cân")
else:
    print("Tam giác thường")
```

### W3A12 – Kiểm tra năm nhuận

```python
n = int(input())
print("Yes" if n % 400 == 0 or (n % 4 == 0 and n % 100 != 0) else "No")
```

### W3A14 – Giải phương trình ax + b = 0

```python
a, b = map(float, input().split())
if a == 0:
    print("Vô số nghiệm" if b == 0 else "Vô nghiệm")
else:
    print(f"{-b / a:.2f}")
```

### W3A15 – Xếp loại học lực

```python
x = float(input())
if x >= 8:
    print("Giỏi")
elif x >= 6.5:
    print("Khá")
elif x >= 5:
    print("Trung bình")
else:
    print("Yếu")
```

### W3A16 – Làm tròn lên, xuống, gần nhất (không dùng hàm làm tròn)

```python
x = float(input())
n = int(x)
up = n + 1 if x > n else n
down = n - 1 if x < n else n
nearest = int(x + 0.5) if x >= 0 else int(x - 0.5)
print(up, down, nearest)
```

### W3A17 – Kiểm tra cấp số nhân với công bội nguyên

```python
a, b, c, d = map(int, input().split())
if a == 0:
    print(b == c == d == 0)
else:
    q = b // a
    print(b % a == 0 and c == b * q and d == c * q)
```
