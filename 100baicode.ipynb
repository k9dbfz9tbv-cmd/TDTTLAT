#### 0059 Nhập vào 1 mảng số nguyên a có n phần tử. Tìm số nguyên xuất hiện nhiều lần nhất trong mảng (nếu có nhiều số cùng xuất hiện nhiều nhất thì lấy số đầu tiên xuất hiện nhiều nhất)

#### 0060 Nhập vào 1 mảng số nguyên a có n phần tử. Tính trung bình cộng các số chính phương trong mảng

```python
a = list(map(int,input().split()))
cnt=0
res=0
def check(a: int):
    i=0
    while(i*i<=a):
        if (i*i==a):
            return True
        else:
            i+=1
    return False
for i in a:
    if check(i):
        cnt+=1
        res+=i
if cnt!=0:
    print(f"{(res/cnt):.1f}")
else:
    print("-")
```
