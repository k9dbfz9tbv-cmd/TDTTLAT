### W2A3 – ModAndDiv
Nhập a, b; in tổng, hiệu, tích, chia nguyên, chia dư, chia thực.

```python
a, b = map(int, input().split())
print(a + b, a - b, a * b, a // b, a % b, f"{a / b:.2f}", sep="\n")
```

### W2A4 – AverageOfSix
Tính trung bình có trọng số.

```python
a1, b1, c1, a2, b2, a3 = map(float, input().split())
tb = (a1 + b1 + c1 + 2 * (a2 + b2) + 3 * a3) / 10
print(f"{tb:.1f}")
```
